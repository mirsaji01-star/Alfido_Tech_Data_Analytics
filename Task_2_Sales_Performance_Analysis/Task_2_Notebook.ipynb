```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin1")

print("Dataset loaded successfully!")
print("Number of rows and columns:", df.shape)

display(df.head())
```

```python

print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nFirst 5 rows:")
display(df.head())
```

```python

print("Missing values in each column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

print("\nSummary statistics:")
display(df.describe(include="all").T)
```

```python

print("Total Sales:", round(df["Sales"].sum(), 2))

print("Total Profit:", round(df["Profit"].sum(), 2))

print("Total Orders:", df["Order ID"].nunique())

print("Total Customers:", df["Customer ID"].nunique())
```

```python

df.columns = df.columns.str.strip()

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")

for col in ["Sales", "Profit", "Quantity", "Discount"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Data preparation completed!")
print("Missing values:")
print(df.isnull().sum())
```

```python

category_sales = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
)

display(category_sales)

category_sales.plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

category_profit = df.groupby("Category")["Profit"].sum().sort_values(
    ascending=False
)

display(category_profit)

category_profit.plot(kind="bar", figsize=(8, 5))
plt.title("Total Profit by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

df["Order Month"] = df["Order Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Order Month")["Sales"].sum()

plt.figure(figsize=(12, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python

region_sales = df.groupby("Region")["Sales"].sum().sort_values(
    ascending=False
)

display(region_sales)

region_sales.plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(
    ascending=False
)

display(segment_sales)

segment_sales.plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

top_products = df.groupby("Product Name").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum")
).sort_values("Total_Sales", ascending=False).head(10)

display(top_products)
```

```python

discount_analysis = df.groupby("Discount").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Number_of_Records=("Discount", "count")
).sort_index()

display(discount_analysis)

plt.figure(figsize=(8, 5))
plt.scatter(df["Discount"], df["Profit"], alpha=0.4)
plt.title("Relationship Between Discount and Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python

shipping_sales = df.groupby("Ship Mode")["Sales"].sum().sort_values(
    ascending=False
)

display(shipping_sales)

shipping_sales.plot(kind="bar", figsize=(8, 5))
plt.title("Total Sales by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Total Sales")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python

subcategory_sales = df.groupby("Sub-Category")["Sales"].sum().sort_values(
    ascending=False
)

display(subcategory_sales)

subcategory_sales.plot(kind="bar", figsize=(12, 5))
plt.title("Total Sales by Product Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=60, ha="right")
plt.tight_layout()
plt.show()
```

```python

summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Average Sales per Order"
    ],
    "Value": [
        df["Sales"].sum(),
        df["Profit"].sum(),
        df["Order ID"].nunique(),
        df["Customer ID"].nunique(),
        df["Sales"].sum() / df["Order ID"].nunique()
    ]
})

display(summary.round(2))
```
