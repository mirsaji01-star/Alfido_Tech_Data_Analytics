```python
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'], errors='coerce')

print("First Purchase Date:", df['Purchase Date'].min())
print("Last Purchase Date:", df['Purchase Date'].max())
print("Number of Customers:", df['Customer ID'].nunique())
```

```python
# Set reference date
reference_date = df['Purchase Date'].max() + pd.Timedelta(days=1)

# Create RFM table
rfm = df.groupby('Customer ID').agg(
    Recency=('Purchase Date', lambda x: (reference_date - x.max()).days),
    Frequency=('Purchase Date', 'count'),
    Monetary=('Total Purchase Amount', 'sum')
).reset_index()

print("RFM table created successfully!")
print(rfm.head())
```

```python
# Create RFM scores

# Recency: lower number is better
rfm['R_Score'] = pd.qcut(
    rfm['Recency'].rank(method='first'),
    4,
    labels=[4, 3, 2, 1]
).astype(int)

# Frequency: higher number is better
rfm['F_Score'] = pd.qcut(
    rfm['Frequency'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Monetary: higher spending is better
rfm['M_Score'] = pd.qcut(
    rfm['Monetary'].rank(method='first'),
    4,
    labels=[1, 2, 3, 4]
).astype(int)

# Total RFM score
rfm['RFM_Score'] = (
    rfm['R_Score'] +
    rfm['F_Score'] +
    rfm['M_Score']
)

print("RFM scores created successfully!")
print(rfm.head())
```

```python
def segment_customer(row):
    if row['RFM_Score'] >= 10:
        return 'High Value'
    elif row['RFM_Score'] >= 7:
        return 'Loyal'
    elif row['RFM_Score'] >= 5:
        return 'Potential'
    else:
        return 'At Risk'

rfm['Segment'] = rfm.apply(segment_customer, axis=1)

print("Customer segmentation completed!")
print(rfm['Segment'].value_counts())
```

```python
# Calculate number of purchases per customer

purchase_frequency = df.groupby('Customer ID').size()

repeat_customers = (purchase_frequency > 1).sum()
one_time_customers = (purchase_frequency == 1).sum()
total_customers = purchase_frequency.count()

print("Total Customers:", total_customers)
print("One-Time Customers:", one_time_customers)
print("Repeat Customers:", repeat_customers)

repeat_rate = (repeat_customers / total_customers) * 100

print("Repeat Customer Rate:", round(repeat_rate, 2), "%")
```

```python
retention_data = pd.Series({
    'One-Time Customers': one_time_customers,
    'Repeat Customers': repeat_customers
})

plt.figure(figsize=(8,5))

retention_data.plot(kind='bar')

plt.title('Customer Retention Analysis')
plt.xlabel('Customer Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)

plt.show()
```
